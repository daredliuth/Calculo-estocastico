## Bibliotecas

```python
import math
import matplotlib.pyplot as plt
import numpy as np
```

```python
s0 = 4
r = 1/4
sH = 8
sT = 2
```

## Cálculo de X0

```python
def FormulaX0(x):
    return x/1+r

x = np.array(range(0,100))

x0 = FormulaX0(x)
plt.plot(x, x0)
plt.show()
```

# Cálculo de Delta0

```python
def FormulaDelta0(x):
    return ((2*x) * (s0-sH)) / ((sT-sH) * (sT-s0))

delta0 = FormulaDelta0(x)
plt.plot(x, delta0, color='red')
plt.show()
```

```python
plt.plot(x, x0, color='blue')
plt.plot(x, delta0, color='red', linestyle='dashed')
plt.show()
```
